# 04 — Ablation Comparison

Compares the four trained configurations (LSTM vs. GRU decoder × frozen vs. fine-tuned encoder) once training and evaluation have been run.

**Run this after** `train.py` and `evaluate.py` have produced results in `experiments/results/`.

```
import sys, json, os
sys.path.insert(0, '..')
import matplotlib.pyplot as plt
```

## Load results for each config

```
configs = ['lstm_frozen', 'lstm_finetuned', 'gru_frozen', 'gru_finetuned']
results = {}
for cfg in configs:
    path = f'../experiments/results/{cfg}_bleu_beam.json'
    if os.path.exists(path):
        with open(path) as f:
            results[cfg] = json.load(f)
    else:
        print(f'[missing] {path} — run evaluate.py for this config first')
```

## BLEU-4 comparison bar chart

```
if results:
    names = list(results.keys())
    bleu4_scores = [results[c]['average_scores']['bleu4'] for c in names]
    plt.figure(figsize=(7,4))
    plt.bar(names, bleu4_scores, color=['#4c72b0','#4c72b0','#dd8452','#dd8452'])
    plt.ylabel('BLEU-4')
    plt.title('BLEU-4 across ablation configurations')
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.savefig('../outputs/bleu_comparison_chart.png')
    plt.show()
else:
    print('No results yet — train and evaluate at least one config first.')
```

## Full BLEU-1 to BLEU-4 table

```
if results:
    import pandas as pd
    rows = []
    for cfg, r in results.items():
        row = {'config': cfg}
        row.update(r['average_scores'])
        rows.append(row)
    df = pd.DataFrame(rows).set_index('config')
    display(df)
```
