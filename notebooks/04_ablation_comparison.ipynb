## BLEU-4 and CIDEr per config (beam 5, mean ± std over seeds)

```python
order = ['lstm_frozen', 'gru_frozen', 'hybrid_frozen', 'lstm_finetuned', 'gru_finetuned', 'hybrid_finetuned']
sub = df[(df.decoding == 'beam5') & df.config.isin(order)]
g = sub.groupby('config')[['bleu4', 'cider']].agg(['mean', 'std']).reindex([c for c in order if c in set(sub.config)])
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
for ax, m in zip(axes, ['bleu4', 'cider']):
    ax.bar(g.index, g[(m, 'mean')], yerr=g[(m, 'std')], capsize=4, color='#4c72b0')
    ax.set_title(m.upper()); ax.tick_params(axis='x', rotation=30)
plt.tight_layout(); os.makedirs('../outputs', exist_ok=True)
plt.savefig('../outputs/ablation_bleu4_cider.png', dpi=150); plt.show()
```
