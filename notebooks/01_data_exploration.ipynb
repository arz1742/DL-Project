# 01 — Data Exploration

Explore the Flickr8k dataset: image counts, caption lengths, vocabulary size.
Runs entirely on CPU — no GPU needed.

```
import sys, os
sys.path.insert(0, '..')
from src.data_loader import load_captions, split_dataset
import matplotlib.pyplot as plt
```

## Load raw captions

Make sure `data/raw/captions.txt` and `data/raw/Images/` exist before running this (see README setup instructions).

```
captions_path = '../data/raw/captions.txt'
image_to_captions = load_captions(captions_path)
print(f'Total images: {len(image_to_captions)}')
```

## Caption length distribution

```
caption_lengths = [len(c.split()) for caps in image_to_captions.values() for c in caps]
plt.figure(figsize=(8,4))
plt.hist(caption_lengths, bins=30)
plt.xlabel('Caption length (words)')
plt.ylabel('Frequency')
plt.title('Distribution of caption lengths')
plt.show()
print(f'Max length: {max(caption_lengths)}, Mean length: {sum(caption_lengths)/len(caption_lengths):.1f}')
```

## Sample captions for a few images

```
import itertools
for name, caps in itertools.islice(image_to_captions.items(), 3):
    print(name)
    for c in caps:
        print('  -', c)
    print()
```

## Train/val/test split preview

```
train_set, val_set, test_set = split_dataset(image_to_captions)
print(f'Train: {len(train_set)}  Val: {len(val_set)}  Test: {len(test_set)}')
```
